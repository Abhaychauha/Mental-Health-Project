# **Project Summary -**

This project explores the **2014 Mental Health in Tech Survey**, a dataset of 1,259 responses collected from employees (largely in the technology industry) that captures demographics, workplace policies, and personal attitudes toward mental health at work. The goal of this EDA is to understand how mental health is perceived, supported, and disclosed in tech workplaces, and to surface the factors most associated with an employee seeking treatment for a mental health condition.

The dataset contains 27 columns covering respondent demographics (Age, Gender, Country, State), employer characteristics (company size, whether the company is a tech company, remote-work policy), the mental-health support infrastructure employers provide (benefits, care options, wellness programs, seek-help resources, anonymity protections, ease of leave), and the respondent's personal attitudes and disclosure comfort (willingness to discuss mental health with coworkers/supervisors, fear of negative consequences, whether they have a family history of mental illness, and whether they have sought treatment).

The data required substantial cleaning before analysis. The `Age` column contained clearly invalid entries (negative values and values in the billions), which were filtered to a realistic working-age range. The `Gender` column had 49 unique free-text responses due to it being an open text field, which were consolidated into three consistent categories: Male, Female, and Other. Columns such as `state` (only applicable to US respondents), `self_employed`, and `work_interfere` contained missing values that were handled contextually — `self_employed` nulls were treated as "No", and `work_interfere` nulls (respondents without a mental health condition) were labeled as "Not applicable". The largely-empty free-text `comments` column was dropped from the numerical analysis.

After cleaning, the EDA moves through univariate analysis (who responded — age, gender, country, company size), then bivariate/multivariate analysis connecting workplace support factors (benefits, care options, anonymity, leave ease) and personal factors (family history, remote work) to the key outcome variable: whether the respondent sought treatment. Fifteen visualizations — including bar charts, stacked/grouped comparisons, box plots, a correlation heatmap, and a pair plot — are used to tell this story, each paired with a rationale for the chart choice and the business insight it produces.

Key patterns that emerge include a strong association between family history of mental illness and treatment-seeking, a clear link between employer-provided benefits/care-option clarity and treatment rates, and workplace-size and geography effects on both support availability and reported attitudes. These findings are translated into concrete recommendations for tech employers on how to design mental-health support programs that measurably increase help-seeking behavior, reduce stigma, and protect employee wellbeing — directly relevant to HR and People Ops teams building or auditing mental-health benefits programs.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df_clean = df.copy()

# 1. Parse Timestamp
df_clean["Timestamp"] = pd.to_datetime(df_clean["Timestamp"])

# 2. Clean Age: keep a realistic working-age range (18-75), fix impossible values
df_clean.loc[(df_clean["Age"] < 18) | (df_clean["Age"] > 75), "Age"] = np.nan
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Age"] = df_clean["Age"].astype(int)

# 3. Create Age Groups for easier grouping
bins = [17, 24, 29, 34, 39, 44, 49, 75]
labels = ["18-24", "25-29", "30-34", "35-39", "40-44", "45-49", "50+"]
df_clean["Age_Group"] = pd.cut(df_clean["Age"], bins=bins, labels=labels)

# 4. Clean Gender into 3 consistent categories: Male, Female, Other
def clean_gender(g):
    g = str(g).strip().lower()
    male_terms = ["male", "m", "man", "cis male", "cis man", "mail", "malr", "maile",
                  "make", "mal", "msle", "guy (-ish) ^_^", "male-ish", "male (cis)",
                  "male leaning androgynous", "ostensibly male, unsure what that really means",
                  "something kinda male?"]
    female_terms = ["female", "f", "woman", "cis female", "cis-female/femme", "femake",
                     "female (cis)", "female (trans)", "femail", "trans-female", "trans woman"]
    if g in male_terms:
        return "Male"
    elif g in female_terms:
        return "Female"
    else:
        return "Other"

df_clean["Gender_Clean"] = df_clean["Gender"].apply(clean_gender)

# 5. Handle missing values contextually
df_clean["self_employed"] = df_clean["self_employed"].fillna("No")
df_clean["work_interfere"] = df_clean["work_interfere"].fillna("Not applicable")
df_clean["state"] = df_clean["state"].fillna("Not in US")

# 6. Drop the mostly-empty free text comments column (not useful for quantitative EDA)
df_clean.drop(columns=["comments"], inplace=True)

print("Shape after cleaning:", df_clean.shape)
df_clean.head()
````

### What all manipulations have you done and insights you found?

- **Age** was cleaned by removing invalid entries (negative numbers, absurd values in the billions) and clipping to a realistic 18–75 working-age range; remaining nulls were filled with the median age. An `Age_Group` bucket column was also created for easier comparison across charts.
- **Gender** was consolidated from 49 messy free-text variants down to **Male / Female / Other** using a mapping function, making group comparisons meaningful.
- Missing values were filled **contextually** rather than dropped: `self_employed` nulls → "No" (majority class), `work_interfere` nulls → "Not applicable" (these are respondents without a mental health condition, so the question doesn't apply to them), and `state` nulls → "Not in US" (since state only applies to US respondents).
- The `comments` column was dropped since it was ~87% empty free text and not suited for quantitative analysis.
- These steps ensure every remaining column is complete and consistently coded, which is required before building reliable, comparable visualizations.

## **5. Solution to Business Objective**

#### What do you suggest the client to achieve Business Objective ?
Explain Briefly.

Based on the EDA, tech employers should focus on three concrete levers to improve mental health outcomes and treatment-seeking among employees:

1. **Close the awareness gap, not just the benefits gap.** The data shows that not knowing about benefits or care options is associated with *lower* treatment-seeking than even having no benefits at all. Companies should run recurring, clear communication (onboarding, intranet, manager reminders) about exactly what mental health support is available and how to access it — this is likely the highest-ROI intervention identified.
2. **Invest in manager-level trust and training.** Employees are more comfortable discussing mental health with coworkers than with direct supervisors. Training managers to respond supportively and confidentially to disclosures can reduce this gap and encourage earlier help-seeking, especially since supervisors often control access to accommodations and leave.
3. **Simplify and clearly document mental health leave.** A large share of respondents don't know how easy it would be to take leave for a mental health condition. Publishing an unambiguous, easy-to-find leave policy (ideally matching the ease of physical-health leave) removes a major source of hesitation.

Because treatment-seeking is roughly evenly split across age, remote-work status, and gender in this sample, these interventions should be applied **broadly across the whole workforce** rather than narrowly targeted, while paying special attention to employees with a family history of mental illness, who are both the most likely to need support and the most predictive group to reach proactively.

# **Conclusion**

This EDA of the 2014 Mental Health in Tech Survey shows that mental health challenges are common and evenly distributed across the tech workforce — roughly half of respondents have sought treatment, largely independent of age, remote-work status, or company size. The strongest predictors of treatment-seeking were **not demographic** but **support-related and personal-risk-related**: having a family history of mental illness, and — critically — **knowing** what mental health benefits and care options an employer provides. Employees who were simply unaware of their support options sought treatment at the lowest rates in the dataset, even lower than those with no benefits at all, highlighting that clear communication of existing support may matter as much as the support itself. Combined with the finding that employees are less comfortable disclosing mental health issues to supervisors than to coworkers, and that many don't know how easy it is to take mental health leave, this analysis points to a clear, low-cost, high-impact action plan for tech employers: communicate benefits clearly and repeatedly, train managers to handle disclosures well, and simplify and publicize leave policies. These changes target the specific gaps — awareness and trust — that this data shows are holding back mental health support from reaching the employees who need it.
